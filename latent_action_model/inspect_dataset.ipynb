```python
import sys
import torch
from pathlib import Path

# 添加项目路径
project_root = Path.cwd().parent
sys.path.append(str(project_root))

from genie.dataset import LightningOpenX
```

```python
# 创建数据模块
data_module = LightningOpenX(
    data_root="/data/home/jlchen/datasets",
    data_mix="droid_bridge",
    batch_size=6,
    resolution=256,
    num_frames=16,
    episodic=False,
    shuffle_buffer_size=1024,
    image_aug=False,
)

# 设置数据集
data_module.setup("fit")

# 获取数据加载器
train_loader = data_module.train_dataloader()
print(f"✅ 数据加载器创建成功")
```

```python
# 获取一个批次的数据
batch = next(iter(train_loader))

print("=== 批次数据结构 ===")
for key, value in batch.items():
    if isinstance(value, torch.Tensor):
        print(f"{key}: {value.shape} | dtype: {value.dtype}")
    elif isinstance(value, list):
        print(f"{key}: List[{len(value)}] | 类型: {type(value[0]) if value else 'Empty'}")
    else:
        print(f"{key}: {type(value)}")
```

```python
# 详细查看视频数据
if 'videos' in batch:
    videos = batch['videos']
    print(f"🎥 视频数据详情:")
    print(f"   形状: {videos.shape}")  # [batch_size, 2, channels, height, width]
    print(f"   数值范围: [{videos.min():.4f}, {videos.max():.4f}]")
    
    # 检查初始帧和目标帧的差异
    batch_size, num_frames, channels, height, width = videos.shape
    print(f"   批次大小: {batch_size}, 帧数: {num_frames} (初始帧+目标帧)")
    print(f"   图像尺寸: {channels}×{height}×{width}")
    
    for i in range(min(10, batch_size)):
        frame_diff = torch.abs(videos[i, 1] - videos[i, 0]).mean()
        print(f"   样本{i} 帧间差异: {frame_diff:.6f}")
```

```python
# 查看任务指令
if 'task_instruction' in batch:
    instructions = batch['task_instruction']
    print(f"📝 任务指令 (共{len(instructions)}条):")
    for i, instruction in enumerate(instructions):
        print(f"   [{i}]: {instruction}")
    
    print(f"\n   唯一指令数: {len(set(instructions))}")
```

```python
# 📊 英文版可视化函数 (兼容无中文字体系统)
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle
import matplotlib.patches as patches

def visualize_batch_samples_en(batch, num_samples=10, figsize=(20, 12)):
    """
    Visualize batch samples with English labels (compatible with systems without Chinese fonts)
    """
    videos = batch['videos']
    instructions = batch['task_instruction'] 
    actions = batch['action']
    dataset_names = batch['dataset_names']
    
    batch_size = min(num_samples, videos.shape[0])
    
    # Create subplots
    fig, axes = plt.subplots(batch_size, 2, figsize=figsize)
    if batch_size == 1:
        axes = axes.reshape(1, -1)
    
    for i in range(batch_size):
        # Get initial and target frames
        initial_frame = videos[i, 0].permute(1, 2, 0).numpy()  # [C,H,W] -> [H,W,C]
        target_frame = videos[i, 1].permute(1, 2, 0).numpy()
        
        # Ensure pixel values are in [0,1] range
        initial_frame = np.clip(initial_frame, 0, 1)
        target_frame = np.clip(target_frame, 0, 1)
        
        # Display initial frame
        axes[i, 0].imshow(initial_frame)
        axes[i, 0].set_title(f'Sample {i} - Initial Frame', fontsize=12, pad=10)
        axes[i, 0].axis('off')
        
        # Display target frame
        axes[i, 1].imshow(target_frame)
        axes[i, 1].set_title(f'Sample {i} - Target Frame', fontsize=12, pad=10)
        axes[i, 1].axis('off')
        
        # Add task instruction and action info below images
        instruction_text = instructions[i]
        dataset_name = dataset_names[i].decode() if isinstance(dataset_names[i], bytes) else dataset_names[i]
        
        # Action vectors (show initial and target actions)
        init_action = actions[i, 0].numpy()
        target_action = actions[i, 1].numpy()
        
        # Calculate frame difference
        frame_diff = torch.abs(videos[i, 1] - videos[i, 0]).mean().item()
        
        # info_text = f"Instruction: {instruction_text}\\n"
        # info_text += f"Dataset: {dataset_name}\\n"
        # info_text += f"Frame Diff: {frame_diff:.4f}\\n"
        # info_text += f"Init Action: [{', '.join([f'{x:.2f}' for x in init_action[:4]])}...]\\n"
        # info_text += f"Target Action: [{', '.join([f'{x:.2f}' for x in target_action[:4]])}...]"
        
        # # Add text below images
        # fig.text(0.02 + (i % 2) * 0.5, 0.95 - (i // 2) * 0.25 - 0.15, info_text, 
        #         fontsize=10, verticalalignment='top', 
        #         bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.7))
    
    plt.tight_layout()
    plt.subplots_adjust(top=0.85)
    plt.suptitle('🤖 VLA Dataset Sample Visualization', fontsize=16, y=0.98)
    plt.show()

def visualize_action_analysis_en(batch, figsize=(18, 10)):
    """
    Visualize action data analysis with English labels
    """
    actions = batch['action']  # [batch_size, 2, 7]
    dataset_names = batch['dataset_names']
    
    fig = plt.figure(figsize=figsize)
    
    # 1. Action dimension heatmap
    ax1 = plt.subplot(2, 3, 1)
    init_actions = actions[:, 0, :].numpy()  # [batch_size, 7]
    im = ax1.imshow(init_actions.T, aspect='auto', cmap='RdBu_r', vmin=-1, vmax=1)
    ax1.set_title('Initial Action Heatmap\\n(Dimension × Sample)', fontsize=12)
    ax1.set_xlabel('Sample Index')
    ax1.set_ylabel('Action Dimension')
    ax1.set_yticks(range(7))
    ax1.set_yticklabels([f'Dim{i}' for i in range(7)])
    plt.colorbar(im, ax=ax1, shrink=0.8)
    
    # 2. Target action heatmap
    ax2 = plt.subplot(2, 3, 2)
    target_actions = actions[:, 1, :].numpy()  # [batch_size, 7]
    im2 = ax2.imshow(target_actions.T, aspect='auto', cmap='RdBu_r', vmin=-1, vmax=1)
    ax2.set_title('Target Action Heatmap\\n(Dimension × Sample)', fontsize=12)
    ax2.set_xlabel('Sample Index')
    ax2.set_ylabel('Action Dimension')
    ax2.set_yticks(range(7))
    ax2.set_yticklabels([f'Dim{i}' for i in range(7)])
    plt.colorbar(im2, ax=ax2, shrink=0.8)
    
    # 3. Box plot for each dimension
    ax3 = plt.subplot(2, 3, 3)
    action_data_for_box = [init_actions[:, i] for i in range(7)]
    box_plot = ax3.boxplot(action_data_for_box, labels=[f'Dim{i}' for i in range(7)])
    ax3.set_title('Action Distribution by Dimension', fontsize=12)
    ax3.set_ylabel('Action Value')
    ax3.set_ylim(-1.1, 1.1)
    ax3.grid(True, alpha=0.3)
    
    # 4. Dataset source pie chart
    ax4 = plt.subplot(2, 3, 4)
    dataset_counts = {}
    for name in dataset_names:
        name_str = name.decode() if isinstance(name, bytes) else str(name)
        dataset_counts[name_str] = dataset_counts.get(name_str, 0) + 1
    
    wedges, texts, autotexts = ax4.pie(dataset_counts.values(), 
                                      labels=dataset_counts.keys(), 
                                      autopct='%1.1f%%',
                                      startangle=90)
    ax4.set_title('Dataset Distribution', fontsize=12)
    
    # 5. Action change magnitude analysis
    ax5 = plt.subplot(2, 3, 5)
    action_changes = np.abs(target_actions - init_actions)  # [batch_size, 7]
    mean_changes = action_changes.mean(axis=0)
    std_changes = action_changes.std(axis=0)
    
    x_pos = range(7)
    bars = ax5.bar(x_pos, mean_changes, yerr=std_changes, capsize=5, 
                  alpha=0.7, color='skyblue', edgecolor='navy')
    ax5.set_title('Action Change Magnitude\\n(Initial → Target)', fontsize=12)
    ax5.set_xlabel('Action Dimension')
    ax5.set_ylabel('Average Change')
    ax5.set_xticks(x_pos)
    ax5.set_xticklabels([f'Dim{i}' for i in range(7)])
    ax5.grid(True, alpha=0.3)
    
    # # Add value labels
    # for i, (bar, mean_val) in enumerate(zip(bars, mean_changes)):
    #     ax5.text(bar.get_x() + bar.get_width()/2., bar.get_height() + std_changes[i] + 0.01,
    #             f'{mean_val:.3f}', ha='center', va='bottom', fontsize=9)
    
    # 6. Action trajectory visualization (first 3 dimensions)
    ax6 = plt.subplot(2, 3, 6)
    for i in range(min(5, len(actions))):  # Show first 5 samples
        for dim in range(3):  # Only show first 3 dimensions
            ax6.plot([0, 1], [init_actions[i, dim], target_actions[i, dim]], 
                    marker='o', linewidth=2, alpha=0.7,
                    label=f'Sample{i}-Dim{dim}' if i < 2 else "")
    
    ax6.set_title('Action Trajectories (First 3 Dim)', fontsize=12)
    ax6.set_xlabel('Timestep (0:Initial, 1:Target)')
    ax6.set_ylabel('Action Value')
    ax6.set_xticks([0, 1])
    ax6.set_xticklabels(['Initial', 'Target'])
    ax6.grid(True, alpha=0.3)
    ax6.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=8)
    
    plt.tight_layout()
    plt.suptitle('🤖 VLA Action Data Analysis', fontsize=16, y=0.98)
    plt.show()
    plt.close()

```
